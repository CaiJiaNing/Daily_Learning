```python
# Question 3
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

# 奇数（元素 % 2 == 1）
odd = arr[arr % 2 == 1]
"""
mask = arr % 2 == 1 --> mask = [True, False, True, False, True, False]
odd = arr[mask] --> odd = [1, 3, 5] 只取出mask为True的元素
"""
# 偶数（元素 % 2 == 0）
even = arr[arr % 2 == 0]
"""
mask = arr % 2 == 0 --> mask = [False, True, False, True, False, True]
even = arr[mask] --> even = [2, 4, 6] 只取出mask为True的元素
"""

print(odd)   # [1 3 5]
print(even)  # [2 4 6]
```

```python
# Question 4
import numpy as np

arr2d = np.array([[-3, -2, -1, 0, 1, 2, 3]])

# 大于 0
gt0 = arr2d[arr2d > 0]

# 大于等于 0
ge0 = arr2d[arr2d >= 0]

# 小于 2
lt2 = arr2d[arr2d < 2]

# 小于等于 2
le2 = arr2d[arr2d <= 2]

print(gt0)  # [1 2 3]
print(ge0)  # [0 1 2 3]
print(lt2)  # [-3 -2 -1  0  1]
print(le2)  # [-3 -2 -1  0  1  2]
```

```python
# Question 5
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])
print(arr.shape)        # (6,)
# Transpose
arr_T = arr.T          
# Or using np.transpose(arr)

print(arr_T)           # [1 2 3 4 5 6]
print(arr_T.shape)     # (6,)

"""
Numpy中的一维数组没有真正的转置操作
因为它们没有行和列的概念
转置操作主要适用于二维及以上的数组
在一维数组中，转置不会改变数组的形状或内容。
"""
```

```python
# Question 6
import numpy as np

arr2d = np.array([[-3, -2, -1, 0, 1, 2, 3]])
print(arr2d.shape)      # (1, 7)

arr2d_T = arr2d.T      
# Or using np.transpose(arr2d)

print(arr2d_T)
# Output:
# [[-3]
#  [-2]
#  [-1]
#  [ 0]
#  [ 1]
#  [ 2]
#  [ 3]]
print(arr2d_T.shape)   # (7, 1)
```
