```python
"""
Matploblib: Definition of colors
"""

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100) # Generate 100 points between 0 and 10
y = np.sin(x)               # Compute the sine of each point 

plt.plot(x, y, color=(0.1, 0.2, 0.5))           # RGB (Red, Green, Blue)
plt.plot(x, y+1, color=(0.1, 0.2, 0.5, 0.3))    # RGBA (Red, Green, Blue, Alpha), where alpha is the transparency level (0.0 to 1.0)

plt.show()  # Display the plot
```

```python
# Hexadecimal color code
plt.plot(x, y, color = '#0f0f0f')    # 6 digits
plt.plot(x, y, color = '#0f0f0f80')  # 8 digits(include transparency)
plt.plot(x, y, color = '#abc')   # Abbreviation, equivalent to '#aabbcc'

plt.show()
```

```python
# Light grey
plt.plot(x, y, color = '0.8')
```

```python
"""
b 蓝色
g 绿色
r 红色
c 青色
m 品红
y 黄色
k 黑色
w 白色
"""
plt.plot(x, y, 'r')
plt.plot(x, y, color = 'g')
```

```python
# X11/CSS4
plt.plot(x, y, color = 'aquamarine')
plt.plot(x, y, color = 'mediumseagreen')
```

```python
# xkcd
plt.plot(x, y, color = 'xkcd:sky blue')
plt.plot(x, y, color = 'eggshell')
```

```python
# Tableau
plt.plot(x, y, color = 'tab:blue')
plt.plot(x, y, color = 'tab:orange')
```

```python
# CN
plt.plot(x, y, color = 'C0') 
```
